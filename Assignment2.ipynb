```python
#Write a function for basic math operations like add multiply substract divide and use this in your program,
#  take 2 number input from user.
num1= int(input("Enter first number"))
num2= int(input("Enter second number"))
print(f"The first number is : {num1}")
print(f"The second number is : {num2}")
add= num1 + num2
print(f"The added result is : {add}")
sub= num1 - num2
print(f"The subtraction result is: {sub}")
multiply= num1 * num2
print(f"The product of number is: {multiply}")
div= num1 / num2
print(f"The division of number is: {div}")

```

```python
# Write a program to check Palindrome Number
str1= input("Enter the palindrome number to check")
print(f"The entered number is: {str1}")
str2= str1[::-1]
if(str1==str2):
    print("The given number is a palindrome")
else:
    print("Try another number.")
```
