```python
#In your last program where you find the total and percentage of a student's marks of 5 subject,
#  find the grade of the student using conditional statement.
# Eg. grade 'A' if percentage is greator than or equals to 60, 'B' for  percentage is greator
# than or equals to 50 and less than 60,  'C' for  percentage is greator than or equals to 40 and less than 50,
#  'D' for  percentage is greator than or equals to 33 and less than 40, otherwise 'Fail'
str1= str(input("The name of the student is :  "))
cla1= str(input("Class of student is :  "))
print(str1)
print(cla1)
m1= int(input("Marks of first subject is: "))
m2= int(input("Marks of 2nd subject is: "))
m3= int(input("Marks of 3rd subject is: "))
m4= int(input("Marks of 4th subject is: "))
m5= int(input("Marks of 5th subject is: "))
marks = m1+m2+m3+m4+m5
print(marks)
p= marks/5
print(p)
if(p>60):
    print("Student has got 'A'")
elif(p<60 and p>=50):
    print("Student has got 'B'")
elif(p<50 and p>=40):
    print("Student has got 'C'")
elif(p<40 and p>=33):
    print("Student has got 'D'")
elif(p<33):
    print("Student has failed")
```
