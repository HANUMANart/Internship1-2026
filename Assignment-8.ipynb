2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 

>Perform an inner merge on this common column and display the resulting DataFrame.
>Perform a left join of df1 and df2 on the 'ID' column. Explain how missing values are handled in the resulting DataFrame. Right Join and Index-Based Join.
>Perform a right join using pd.merge() on a common column, then perform a join using df.join() based on the index. Compare the results. Merging with Multiple Keys.


```python
import pandas as pd
df1= pd.DataFrame({
    "s_name": ["Hanu","Hitku","Utk","Sheksha"],
    "Id": [2, 4,6, 8],
    "Income":[10000, 15000, 32000, 8000],
    "Index": [0,1,2,3]
})
df2= pd.DataFrame({
    "s_name": ["Hitku","Manu","Utk","Sha"],
    "Id": [2, 4,7, 8],
    "Income":[1000, 12000, 2000, 800],
    "Role": ["Manager", "Seniour_manager", "HR", "HOD"],
    "Index":[1,2,3,4]
})
res= df1.merge(df2, on= "Id", how="inner")
print(res,"\n")
res1= df1.merge(df2, on= "Id", how="right")
print(res1,"\n")
res2= df1.merge(df2, on= "Id", how="left")
print(res2,"\n")
res3= df1.merge(df2,  left_on=None, right_on=None,
left_index=True, right_index=True, sort=False)
print(res3,"\n")
res4= pd.merge(df1,df2,on= "Id", how= "right")
print(res4)
result = df1.join(df2, lsuffix='_left', rsuffix='_right')
print(result)
res5= df2.merge(df1, on=["Id", "s_name"])
print(res5)
```

3) Create three DataFrames. Vertically concatenate two of them using pd.concat(), then merge the resulting DataFrame with the third DataFrame on a common key. T Understand join() vs. merge().

Also Explain the primary differences between df.join() and pd.merge(),

```python
students = pd.DataFrame({
    'stu_id': [1, 2, 3, 4, 5],
    'name': ['Aman', 'Hitesh', 'Parvez', 'Anmol', 'Adarsh'],
    'age': [18, 20, 25, 17, 22]
})
books = pd.DataFrame({
    'stu_id': [4, 3, 2, 5],
    'book_name': ['The Hobbit', 'The Martian', 'Gone Girl', 'The Kite Runner'],
    'date_issued': ['2026-01-23', '2026-02-10', '2026-02-10', '2026-01-15']
})
dept = pd.DataFrame({
    'stu_id': [1, 2, 3, 4, 5],
    'department': ['CSE', 'ECE', 'ME', 'CSE', 'EE'],
    'semester': [5, 5, 6, 4, 5]
})
vertical = pd.concat(
    [students, books, dept],
    axis=0,
    ignore_index=True
)
print(vertical)
# When concatenating vertically, pandas creates the union of all columns.
# For rows coming from one DataFrame, columns that don't exist in that DataFrame
# are filled with NaN.
print("_______________________________________________________________")
new_merge = pd.merge(vertical, dept)
print(new_merge)
```
